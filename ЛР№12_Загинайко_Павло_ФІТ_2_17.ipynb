```python
from scipy import integrate
import numpy as np
```

изображение.png

```python
a = 0.4
b = 1.2
n = 10
h = (b - a) / n  # Крок

def f1(x):
    return 1 / np.sqrt(x + 0.5)

# Метод лівих прямокутників
def left_rec(f1, a, b):
    sum_val = 0
    for i in range(0, n):
        sum_val += f1(a + i * h)
    return sum_val * h

print("Лівий прямокутник:", round(left_rec(f1, a, b), 5))

# Метод правих прямокутників
def right_rec(f1, a, b):
    sum_val = 0
    for i in range(1, n + 1):
        sum_val += f1(a + i * h)
    return sum_val * h

print("Правий прямокутник:", round(right_rec(f1, a, b), 5))

# Метод середніх прямокутників
def aver_rec(f1, a, b):
    sum_val = 0
    for i in range(0, n):
        sum_val += f1(a + (i + 0.5) * h)
    return sum_val * h

print("Середній прямокутник:", round(aver_rec(f1, a, b), 5))

# Перевірка за допомогою функції scipy
v, err = integrate.quad(f1, a, b)
print("Перевірка за допомогою методу SciPy quad:", round(v, 5))
```

```python
def f(x):
    return np.cos(x) / (x + 2)

# Межі інтегрування
a = 0.4
b = 1.2
n = 8  # Кількість розбиттів
h = (b - a) / n

def simpson_rule(f, a, b, n):
    integr = f(a) + f(b)
    for i in range(1, n):
        k = a + i * h
        if i % 2 == 0:
            integr += 2 * f(k)
        else:
            integr += 4 * f(k)
    integr *= h / 3
    return integr

# Виводимо результат
result_simpson = simpson_rule(f, a, b, n)
print("Метод Сімпсона:", round(result_simpson, 3))

# Перевірка за допомогою вбудованої функції quad
v, err = integrate.quad(f, a, b)
print("Перевірка для методу Сімпсона = ", round(v, 3))
```

```python
def f(x):
    return 1 / np.sqrt(x**2 - 3)

# Задаємо межі інтегрування
a = 2.1
b = 3.6
n = 20 # Кількість розбиттів

def trapezoidal_rule(f, a, b, n):
    h = (b - a) / n
    total_sum = 0.5 * (f(a) + f(b))
    x = a

    for i in range(1, n):
        x += h
        total_sum += f(x)

    integral = h * total_sum
    return integral

# Виводимо результат
integral_trap = trapezoidal_rule(f, a, b, n)
print(f"Результат методом трапецій (n={n}):", round(integral_trap, 5))

# Перевірка через вбудовану функцію SciPy
v, err = integrate.quad(f, a, b)
print("Точне значення (SciPy quad):", round(v, 5))
print("Різниця:", round(abs(v - integral_trap), 5))
```
